# Business Operations Performance Analysis

## Python / Pandas Analysis

This notebook analyzes a simulated business operations dataset to identify revenue trends, operational performance, cancellation patterns, and delivery performance.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../business_operations.csv')
df.head()
```

## 1. Data Overview

Review the number of records, columns, data types, and missing values before beginning the analysis.

```python
print(f'Rows: {df.shape[0]:,}')
print(f'Columns: {df.shape[1]:,}')

print('\nData types:')
print(df.dtypes)

print('\nMissing values:')
print(df.isna().sum())
```

## 2. Key Performance Indicators

Calculate the core KPIs defined in the Business Requirements Document.

```python
total_revenue = df['Revenue'].sum()
total_orders = df['Order_ID'].nunique()
average_order_value = df['Revenue'].mean()
cancellation_rate = (df['Order_Status'].eq('Cancelled').mean()) * 100
average_processing_time = df['Processing_Time_Days'].mean()

delivery_orders = df[df['On_Time_Flag'].isin(['Yes', 'No'])]
on_time_rate = delivery_orders['On_Time_Flag'].eq('Yes').mean() * 100

kpis = pd.DataFrame({
    'KPI': [
        'Total Revenue', 'Total Orders', 'Average Order Value',
        'Cancellation Rate', 'Average Processing Time (Days)',
        'On-Time Delivery Rate'
    ],
    'Value': [
        total_revenue, total_orders, average_order_value,
        cancellation_rate, average_processing_time, on_time_rate
    ]
})

kpis
```

## 3. Revenue by Product Category

Identify which product categories contribute the most revenue.

```python
category_revenue = (
    df.groupby('Product_Category')['Revenue']
      .agg(['sum', 'count', 'mean'])
      .sort_values('sum', ascending=False)
)

category_revenue
```

```python
category_revenue['sum'].plot(kind='bar')
plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Revenue ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 4. Revenue by Region

```python
region_revenue = (
    df.groupby('Region')['Revenue']
      .agg(['sum', 'count', 'mean'])
      .sort_values('sum', ascending=False)
)

region_revenue
```

## 5. Cancellation Analysis

Examine the reasons orders were cancelled so the business can identify potential improvement opportunities.

```python
cancellation_analysis = (
    df[df['Order_Status'] == 'Cancelled']
    .groupby('Cancellation_Reason')
    .size()
    .sort_values(ascending=False)
)

cancellation_analysis
```

```python
cancellation_analysis.plot(kind='bar')
plt.title('Cancellation Reasons')
plt.xlabel('Cancellation Reason')
plt.ylabel('Number of Cancelled Orders')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6. Processing Time by Product Category

```python
processing_by_category = (
    df.groupby('Product_Category')['Processing_Time_Days']
      .mean()
      .sort_values(ascending=False)
)

processing_by_category
```

```python
processing_by_category.plot(kind='bar')
plt.title('Average Processing Time by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Average Processing Time (Days)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 7. Monthly Revenue Trend

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

monthly_revenue = (
    df.groupby(df['Order_Date'].dt.to_period('M'))['Revenue']
      .sum()
)

monthly_revenue
```

```python
monthly_revenue.plot(kind='line', marker='o')
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
